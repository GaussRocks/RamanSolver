# **Método Newton-Raphson Multidimensional**

Considere a figura a seguir para um interpretação geométrica

![image](newton_raphson.png)

Do cálculo, sabe que a derivada direcional de uma furva $f(\bold{x})$ na diração $\bold{u}$ tal que $|\bold{u}|=1$ é dado por $D_u = \bold{u} \bold{\nabla}f$.
Sendo assim, da geometria, pode escrever que 
$$
\frac{f(\bold{x_k})}{|\bold{x_k}-\bold{x_{k+1}}|} = \frac{\bold{x_k}-\bold{x_{k+1}}}{|\bold{x_k}-\bold{x_{k+1}}|}\bold{\nabla}f

$$

então 

$$
f(\bold{x_k}) = (\bold{x_k}-\bold{x_{k+1}})\bold{\nabla}f
$$

Extendendo o conceito, considere agora o campo vetorial 

$$
\bold{F} = \begin{bmatrix} f_1 & f_2 & ... & f_n \end{bmatrix}^T
$$

Podemos escrever 

$$
\bold{F}(\bold{x_k}) = \begin{bmatrix} (\bold{x_k}-\bold{x_{k+1}})\bold{\nabla}f_1 & (\bold{x_k}-\bold{x_{k+1}})\bold{\nabla}f_2 & ... & (\bold{x_k}-\bold{x_{k+1}})\bold{\nabla}f_n \end{bmatrix}^T
$$

ou 

$$
\bold{F}(\bold{x_k}) = 
\begin{bmatrix}
\bold{\nabla}f_1 \\ \bold{\nabla}f_2 \\ \vdots \\ \bold{\nabla}f_n
\end{bmatrix}
\begin{bmatrix}
x_k^1 - x_{k+1}^1 & x_k^2 - x_{k+1}^2 & ... & x_k^m - x_{k+1}^m
\end{bmatrix}^T
$$

por fim

$$
\bold{F}(\bold{x_k}) = 
\begin{bmatrix} 
\frac{\partial f_1}{\partial x^1} & \frac{\partial f_1}{\partial x^2} & \cdots & \frac{\partial f_1}{\partial x^m} \\ 
\frac{\partial f_2}{\partial x^1} & \frac{\partial f_2}{\partial x^2} & \cdots & \frac{\partial f_2}{\partial x^m} \\ 
\vdots & \vdots & \ddots & \vdots \\ 
\frac{\partial f_n}{\partial x^1} & \frac{\partial f_n}{\partial x^2} & \cdots & \frac{\partial f_n}{\partial x^m} 
\end{bmatrix}
\begin{bmatrix}
x_k^1 - x_{k+1}^1 \\ x_k^2 - x_{k+1}^2 \\ \vdots \\ x_k^m - x_{k+1}^m
\end{bmatrix} =
\bold{J}(\bold{x_k}) (\bold{x_k} - \bold{x_{k+1}})
$$

em que $\bold{J}(\bold{x_k})$ é a matriz Jacobiana de $\bold{F}(\bold{x_k})$

Sendo assim, o método de Newton-Raphson multidimensional é dado por

$$
\bold{x_{k+1}} = \bold{x_k} -\bold{J}^{-1}(\bold{x_k}) \bold{F}(\bold{x_k}) 
$$
